# Program 7 - Lexical Analyzer using Flex
Tokenizes C-like source code into keywords, numbers, identifiers, operators and symbols.

```python
%%writefile scanner.l
%{
#include <stdio.h>

int total_tokens = 0;

#define EMIT(kind) do { total_tokens++; printf("%-12s : %s\n", kind, yytext); } while (0)
%}

%option noyywrap

DIGIT   [0-9]
LETTER  [a-zA-Z_]

%%

"int"|"float"|"char"|"double"|"if"|"else"|"while"|"for"|"return"|"void"   { EMIT("RESERVED"); }

{DIGIT}+"."{DIGIT}+                { EMIT("FLOAT"); }
{DIGIT}+                           { EMIT("INTEGER"); }
{LETTER}({LETTER}|{DIGIT})*        { EMIT("IDENTIFIER"); }
\"[^\"\n]*\"                       { EMIT("STRING"); }

"//".*                             { /* skip single-line comments */ }

"=="|"!="|"<="|">="|"&&"|"||"|"++"|"--"   { EMIT("OPERATOR"); }
[+\-*/%=<>!]                       { EMIT("OPERATOR"); }
[{}();,\[\]]                       { EMIT("SYMBOL"); }

[ \t\r\n]+                         { /* skip whitespace */ }
.                                  { EMIT("UNKNOWN"); }

%%

int main(void)
{
    yylex();
    printf("\nTotal tokens found: %d\n", total_tokens);
    return 0;
}
```
